```python
from sentence_transformers import SentenceTransformer, util

# We are downloading a small, fast embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded successfully!")
```

```python
word = "Apple"
embedding = model.encode(word)
print(embedding[:5])
```

```python
Parisem=model.encode("Paris")
Franceem=model.encode("France")
Romeem=model.encode("Rome")
Italyem=model.encode("Italy")
similarity = util.cos_sim(Romeem,Franceem-Parisem+Italyem)
print(similarity.item())
```

```python
words=['strawberry','banana','car']
embeddings=model.encode(words)
vs_banana=util.cos_sim(embeddings[0],embeddings[1]).item()
vs_car=util.cos_sim(embeddings[0],embeddings[2]).item()
```

```python
print(f"Similarity between Strawberry & Banana: {vs_banana:.4f}")
print(f"Similarity between Strawberry & Car:    {vs_car:.4f}")
```
