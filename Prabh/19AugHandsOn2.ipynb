```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("gpt2")

prompt = "The capital of France is"
inputs = tokenizer(prompt, return_tensors="pt")

# Get the raw math scores (logits) for the next word
with torch.no_grad():
    outputs = model(**inputs)
    next_word_logits = outputs.logits[0, -1, :]

# Convert raw math into percentages (Softmax)
probabilities = torch.nn.functional.softmax(next_word_logits, dim=-1)

# Get the top 5 most likely words
top_5_probs, top_5_indices = torch.topk(probabilities, 5)

print(f"Prompt: '{prompt}'\n")
print("Top 5 Next Word Guesses:")
for prob, idx in zip(top_5_probs, top_5_indices):
    word = tokenizer.decode(idx)
    print(f"- {word:<10} : {prob.item() * 100:.2f}%")
```
