```python
messy_cust['full_name']=messy_cust['full_name'].str.strip().str.title()
```

```python
messy_cust['email']=messy_cust['email'].str.lower()
```

```python
messy_cust.age.describe()
```

```python
q1=messy_cust["age"].quantile(0.25)
q3 = messy_cust["age"].quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(lower_bound, upper_bound)
```

```python
messy_cust["age"] = [
    x if lower_bound <= x <= upper_bound else 54
    for x in messy_cust["age"]
]
```

```python
messy_cust.loc[messy_cust['email'].str.contains('@invalid_domain', na=False),'email'] = 'not_available'
```

```python
messy_cust['account_balance'].isnull().sum()
```

```python
messy_cust.fillna({'account_balance':0}, inplace=True)
```

```python
messy_cust.dtypes
```

```python
messy_cust['account_balance'] = pd.to_numeric(
    messy_cust['account_balance']
      .astype(str)
      .str.replace(r'[\$,]', '', regex=True),
    errors='coerce'
)
messy_cust['account_balance'].dtype
```

```python
messy_cust.fillna({'phone':'0'},inplace=True)
```

```python
def transform_phone(phone_number:str):
    phone_number=phone_number.replace('(','')
    phone_number=phone_number.replace(')','')
    phone_number=phone_number.replace('-','')
    phone_number=phone_number.replace(' ','')
    phone_number=phone_number.replace('.','')
    phone_number=phone_number.strip()
    return int(phone_number)
```

```python
messy_cust['phone'] = messy_cust['phone'].apply(transform_phone)
```
